# 下单时段分布

看下单集中在一天中的哪几个小时、一周中的哪几天。

```python
import pandas as pd
from db import query

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 180)
```

```python
query("""
SELECT HOUR(purchase_ts) AS hour,
       COUNT(*) AS orders
FROM dwd_orders
WHERE purchase_ts IS NOT NULL
GROUP BY hour
ORDER BY hour
""")
```

```python
query("""
SELECT DAYOFWEEK(purchase_ts) AS dow,  -- 1 = 周日
       COUNT(*) AS orders
FROM dwd_orders
WHERE purchase_ts IS NOT NULL
GROUP BY dow
ORDER BY dow
""")
```

下单集中在 10 点到 16 点，周末和工作日差别不大，说明这是 B2C 的常规节奏，不是靠促销脉冲拉起来的。
