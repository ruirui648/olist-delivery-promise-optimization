# 评分分布

评价表按 order_id 聚合过，一个订单可能有多条评价，取平均分。

```python
import pandas as pd
from db import query

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 180)
```

```python
query("""
SELECT review_score,
       COUNT(*) AS orders,
       ROUND(COUNT(*) / SUM(COUNT(*)) OVER () * 100, 2) AS pct
FROM dwd_reviews
GROUP BY review_score
ORDER BY review_score
""")
```

## 差评率最高的类目

只看订单量 300 以上的类目，避免小样本。

```python
query("""
SELECT category,
       COUNT(*) AS orders,
       ROUND(AVG(review_score), 2) AS avg_score,
       ROUND(AVG(review_score <= 2) * 100, 2) AS bad_pct
FROM dws_order_wide
WHERE review_score IS NOT NULL
GROUP BY category
HAVING COUNT(*) >= 300
ORDER BY bad_pct DESC
LIMIT 10
""")
```
