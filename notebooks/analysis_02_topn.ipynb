# 类目 / 卖家 / 州 TOP N

按商品销售额（price 合计，不含运费）排名。

```python
import pandas as pd
from db import query

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 180)
```

## 类目 TOP10

```python
query("""
SELECT category,
       COUNT(DISTINCT order_id) AS orders,
       ROUND(SUM(price), 2) AS gmv
FROM dws_order_wide
GROUP BY category
ORDER BY gmv DESC
LIMIT 10
""")
```

## 卖家 TOP10

```python
query("""
SELECT seller_id,
       COUNT(DISTINCT order_id) AS orders,
       ROUND(SUM(price), 2) AS gmv
FROM dws_order_wide
GROUP BY seller_id
ORDER BY gmv DESC
LIMIT 10
""")
```

## 州 TOP10

```python
query("""
SELECT customer_state,
       COUNT(DISTINCT order_id) AS orders,
       ROUND(SUM(price), 2) AS gmv
FROM dws_order_wide
GROUP BY customer_state
ORDER BY gmv DESC
LIMIT 10
""")
```

SP 一个州占四成左右，和它的人口与经济体量相符。
