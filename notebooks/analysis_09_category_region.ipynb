# 类目 × 区域

各州对不同类目的偏好差异，用于区域选品和备货。

```python
import pandas as pd
from db import query

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 180)
```

```python
df = query("""
SELECT category, customer_state, ROUND(SUM(price), 2) AS gmv
FROM dws_order_wide
GROUP BY category, customer_state
""")
top_cat = df.groupby("category").gmv.sum().nlargest(10).index
top_state = df.groupby("customer_state").gmv.sum().nlargest(8).index
pivot = df[df.category.isin(top_cat) & df.customer_state.isin(top_state)] \
    .pivot(index="category", columns="customer_state", values="gmv") \
    .round(0)
pivot
```

各州的类目结构差别不大，主要还是 SP/RJ/MG 撑量，区域差异体现在绝对规模而不是品类偏好上。
