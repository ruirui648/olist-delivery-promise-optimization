# 类目集中度（帕累托）

用窗口函数算累计占比，看多少个类目撑起 80% 的销售额。

```python
import pandas as pd
from db import query

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 180)
```

```python
df = query("""
SELECT category,
       ROUND(SUM(price), 2) AS gmv,
       ROUND(SUM(SUM(price)) OVER (ORDER BY SUM(price) DESC)
             / (SELECT SUM(price) FROM dws_order_wide) * 100, 2) AS cum_pct
FROM dws_order_wide
GROUP BY category
ORDER BY gmv DESC
""")
df.head(15)
```

```python
n80 = (df.cum_pct < 80).sum() + 1
print(f"共 {len(df)} 个类目，前 {n80} 个贡献 80% 的销售额，占类目数的 {n80/len(df)*100:.1f}%")
```
