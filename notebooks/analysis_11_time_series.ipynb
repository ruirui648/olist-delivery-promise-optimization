# 月度趋势

按月看 GMV 和订单量，再单独看 2017 年黑五那一周。

```python
import pandas as pd
from db import query

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 180)
```

```python
query("""
SELECT DATE_FORMAT(purchase_ts, '%Y-%m') AS month,
       COUNT(DISTINCT order_id) AS orders,
       ROUND(SUM(price), 2) AS gmv,
       ROUND(SUM(price) / COUNT(DISTINCT order_id), 2) AS aov
FROM dws_order_wide
WHERE purchase_ts >= '2017-01-01'
GROUP BY month
ORDER BY month
""")
```

## 黑五周

2017 年黑五在 11 月 24 日，取前后各一周对比。

```python
query("""
SELECT CASE
         WHEN purchase_ts BETWEEN '2017-11-20' AND '2017-11-26' THEN '黑五周'
         WHEN purchase_ts BETWEEN '2017-11-13' AND '2017-11-19' THEN '前一周'
         WHEN purchase_ts BETWEEN '2017-11-27' AND '2017-12-03' THEN '后一周'
       END AS week_label,
       COUNT(DISTINCT order_id) AS orders,
       ROUND(SUM(price), 2) AS gmv
FROM dws_order_wide
WHERE purchase_ts BETWEEN '2017-11-13' AND '2017-12-03'
GROUP BY week_label
""")
```
