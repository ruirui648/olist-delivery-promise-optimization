# 卖家履约表现

看单个卖家的单量、准时率和评分，找出「量不小但履约差」的卖家。

```python
import pandas as pd
from db import query

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 180)
```

```python
query("""
SELECT seller_id,
       COUNT(*) AS orders,
       ROUND(AVG(slack_days < 0) * 100, 2) AS late_pct,
       ROUND(AVG(review_score), 2) AS avg_score,
       ROUND(SUM(price), 2) AS gmv
FROM dws_order_wide
WHERE order_status = 'delivered'
GROUP BY seller_id
HAVING COUNT(*) >= 50
ORDER BY late_pct DESC
LIMIT 10
""")
```

单量 50 以上的卖家里，延迟率高的那几家评分也明显偏低，可以优先约谈或限流。
